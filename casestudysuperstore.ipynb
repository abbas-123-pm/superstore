```python
Sp=pd.read_csv("Sample_Superstore.csv", encoding="latin1")
print(Sp)
```

```python
#1.	What is the size and structure of the dataset?
print("size",Sp.shape)
```

```python
print("structure",Sp.info)
```

```python
#2.	Are there missing values or duplicate records?
print("missing values\n ",Sp.isnull().sum().sum())
```

```python
print("duplicate",Sp.duplicated().sum().sum())
```

```python
#3.	What are the total sales, total profit, and average order-line sales?
print("total sales",Sp['Sales'].sum())
```

```python
print("total profit",Sp['Profit'].sum())
```

```python
print("Average sales",Sp['Sales'].mean())
```

```python
#4.	Which product categories generate the most sales?
sales_category=Sp.groupby('Category')['Sales'].sum()
print(sales_category.sort_values(ascending=False))
```

```python
#5.	Which sub-categories generate the highest and lowest profit?
print(Sp.groupby('Sub-Category')['Profit'].sum().sort_values())
```

```python
#6.	Which region has the highest sales?
print(Sp.groupby('Region')['Sales'].sum().idxmax())
```

```python
#7.	Which customer segment contributes the most sales?
print(Sp.groupby('Segment')['Sales'].sum().idxmax())
```

```python
#8.	How do discounts relate to profit?
print(Sp.groupby('Discount')['Profit'].mean())
```

```python
#9.	How do sales change over time?
Sp['Order Date']=pd.to_datetime(Sp['Order Date'])
print(Sp['Order Date'].dtype)
```
